```python
import torch

# Define the device to use MPS for M1/M2/M3 GPU acceleration
if torch.backends.mps.is_available():
    device = torch.device("mps")
    print("✅ Using Metal Performance Shaders (MPS) for M1 GPU acceleration.")
    print("All models and tensors will be moved to this device.")
else:
    device = torch.device("cpu")
    print("⚠️ MPS not available. Training will run on CPU, which will be slow.")
    
# You will use this 'device' variable for Mask R-CNN training.
# For YOLO, you must use the string 'mps' directly in the training arguments.
```

```python
from ultralytics import YOLO
import os

# --- 1. Define Paths and Model ---

# IMPORTANT: Replace this with the absolute path to your data.yaml file!
DATA_YAML_PATH = '/Users/umairmomin/Desktop/Smart Packaging/dataset/Labeled_data12/data.yaml' 

# Load a pre-trained YOLO model (e.g., 's' small version is a good start)
# For open-vocabulary detection, you could use 'yolov8s-world.pt' 
# For fine-tuning on your 170 images, a standard YOLOv8 model is robust.
MODEL_NAME = 'yolov8s-worldv2.pt'  # 'n' for nano (fastest), 's' for small (best balance)

# --- 2. Load the Model ---
model = YOLO(MODEL_NAME) 

# --- 3. Start Accelerated Training ---
print(f"\n🚀 Starting YOLO training on device: {device.type.upper()}")

# The device='mps' argument is CRITICAL for M1 GPU acceleration
results = model.train(
    data=DATA_YAML_PATH,         # Path to your dataset configuration
    epochs=100,                  # Train for 100 epochs
    imgsz=640,                   # Image size (standard for YOLO)
    batch=16,                    # Adjust based on your M1's memory (e.g., 8, 16, 32)
    name='yolo_packaging_run',   # Name for the saved results folder
    device='mps'                 # <<-- CRITICAL M1 ACCELERATION FLAG
)

print("\n✅ YOLO Training Complete. Weights saved to runs/detect/yolo_packaging_run/weights/best.pt")
```
